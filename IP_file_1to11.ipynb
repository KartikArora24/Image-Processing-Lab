Write a Program to convert Grayscale image to binary image.

● Case 1: Assume Mean Intensity Image as threshold value.

● Case 2: Input the threshold value from the user.

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)

if img is None:
    print("Image not found!")
    exit()

threshold1 = np.mean(img)

binary1 = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        if img[i, j] >= threshold1:
            binary1[i, j] = 255
        else:
            binary1[i, j] = 0

print("Mean Intensity Threshold:", threshold1)

threshold2 = int(input("Enter threshold value (0-255): "))

if threshold2 < 0 or threshold2 > 255:
    print("Invalid threshold!")
    exit()

binary2 = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        if img[i, j] >= threshold2:
            binary2[i, j] = 255
        else:
            binary2[i, j] = 0

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(binary1, cmap='gray')
plt.title("Mean Threshold")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(binary2, cmap='gray')
plt.title("User Threshold")
plt.axis("off")

plt.show()
```

The weightage
is a value between 0 and 1, and sum of all three weightages is equal to 1. (Example:
Weightage of R, G, B is 0.7, 0.2, 0.1, respectively.)

```python
#case 1
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg")

if img is None:
    print("Image not found!")
    exit()

B, G, R = cv2.split(img)

gray = np.zeros((img.shape[0], img.shape[1]), dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        gray[i, j] = int((int(R[i, j]) + int(G[i, j]) + int(B[i, j])) / 3)

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original RGB Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(gray, cmap="gray")
plt.title("Grayscale Image - Mean Average")
plt.axis("off")

plt.show()
```

The input range of intensity values and the desired range of intensity values in the

1. Write a program to enhance the contrast of the input grayscale image
using histogram equalization. Display the original image and equalized
image.

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)

hist = np.zeros(256, dtype=int)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        hist[img[i, j]] += 1

cdf = np.zeros(256, dtype=int)
cdf[0] = hist[0]

for i in range(1, 256):
    cdf[i] = cdf[i-1] + hist[i]

cdf_min = min(cdf[cdf > 0])
total = img.shape[0] * img.shape[1]

mapping = np.zeros(256, dtype=np.uint8)

for i in range(256):
    mapping[i] = int(((cdf[i] - cdf_min) / (total - cdf_min)) * 255)

result = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        result[i, j] = mapping[img[i, j]]

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result, cmap="gray")
plt.title("Histogram Equalized Image")
plt.axis("off")

plt.show()
```

2. Write a program to perform histogram matching of the input image with
that of reference image using histogram equalization

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)
ref = cv2.imread("reference.jpg", cv2.IMREAD_GRAYSCALE)

if img is None or ref is None:
    print("Image not found!")
    exit()

hist1 = np.zeros(256, dtype=int)
hist2 = np.zeros(256, dtype=int)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        hist1[img[i, j]] += 1

for i in range(ref.shape[0]):
    for j in range(ref.shape[1]):
        hist2[ref[i, j]] += 1

cdf1 = np.cumsum(hist1) / img.size
cdf2 = np.cumsum(hist2) / ref.size

mapping = np.zeros(256, dtype=np.uint8)

for i in range(256):
    diff = np.abs(cdf2 - cdf1[i])
    mapping[i] = np.argmin(diff)

result = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        result[i, j] = mapping[img[i, j]]

plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(img, cmap="gray")
plt.title("Input Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(ref, cmap="gray")
plt.title("Reference Image")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(result, cmap="gray")
plt.title("Histogram Matched Image")
plt.axis("off")

plt.show()
```

Write a program to smooth the input image using:
1. Averaging filter (un-weighted)
2. Weighted filter given by h(x,y)= max(|x|.|y|)
3. Gaussian filter

Note: Input the filter size and other parameters such as sigma in case of Gaussian filter,
method of padding (replication, zero padding, etc.) from the user.

```python
#Averaging Filter
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)

n = int(input("Enter filter size (odd): "))
pad = input("Enter padding (zero/replicate): ")

if n <= 0 or n % 2 == 0:
    print("Filter size must be a positive odd number")
    exit()

p = n // 2
h, w = img.shape

padded = np.zeros((h + 2*p, w + 2*p), dtype=np.uint8)

if pad == "replicate":
    for i in range(h + 2*p):
        for j in range(w + 2*p):
            x = min(max(i-p, 0), h-1)
            y = min(max(j-p, 0), w-1)
            padded[i, j] = img[x, y]
else:
    padded[p:p+h, p:p+w] = img

result = np.zeros(img.shape, dtype=np.uint8)

for i in range(h):
    for j in range(w):
        total = 0
        for x in range(n):
            for y in range(n):
                total += int(padded[i+x, j+y])
        result[i, j] = int(total / (n*n))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result, cmap="gray")
plt.title("Averaging Filter")
plt.axis("off")

plt.show()
```

```python
#Weighted Filter

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)

n = int(input("Enter filter size (odd): "))
pad = input("Enter padding (zero/replicate): ")

if n <= 0 or n % 2 == 0:
    print("Filter size must be a positive odd number")
    exit()

p = n // 2
h, w = img.shape

kernel = np.zeros((n, n), dtype=float)

for i in range(n):
    for j in range(n):
        kernel[i, j] = max(abs(i-p), abs(j-p))

kernel = kernel / np.sum(kernel)

padded = np.zeros((h + 2*p, w + 2*p), dtype=np.uint8)

if pad == "replicate":
    for i in range(h + 2*p):
        for j in range(w + 2*p):
            x = min(max(i-p, 0), h-1)
            y = min(max(j-p, 0), w-1)
            padded[i, j] = img[x, y]
else:
    padded[p:p+h, p:p+w] = img

result = np.zeros(img.shape, dtype=np.uint8)

for i in range(h):
    for j in range(w):
        total = 0
        for x in range(n):
            for y in range(n):
                total += padded[i+x, j+y] * kernel[x, y]
        result[i, j] = int(total)

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result, cmap="gray")
plt.title("Weighted Filter")
plt.axis("off")

plt.show()
```

```python
#Gaussian Filter

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", cv2.IMREAD_GRAYSCALE)

n = int(input("Enter filter size (odd): "))
sigma = float(input("Enter sigma: "))
pad = input("Enter padding (zero/replicate): ")

if n <= 0 or n % 2 == 0 or sigma <= 0:
    print("Invalid filter size or sigma")
    exit()

p = n // 2
h, w = img.shape

kernel = np.zeros((n, n), dtype=float)

for i in range(n):
    for j in range(n):
        x = i - p
        y = j - p
        kernel[i, j] = np.exp(-(x*x + y*y)/(2*sigma*sigma))

kernel = kernel / np.sum(kernel)

padded = np.zeros((h + 2*p, w + 2*p), dtype=np.uint8)

if pad == "replicate":
    for i in range(h + 2*p):
        for j in range(w + 2*p):
            x = min(max(i-p, 0), h-1)
            y = min(max(j-p, 0), w-1)
            padded[i, j] = img[x, y]
else:
    padded[p:p+h, p:p+w] = img

result = np.zeros(img.shape, dtype=np.uint8)

for i in range(h):
    for j in range(w):
        total = 0
        for x in range(n):
            for y in range(n):
                total += padded[i+x, j+y] * kernel[x, y]
        result[i, j] = int(total)

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(result, cmap="gray")
plt.title("Gaussian Filter")
plt.axis("off")

plt.show()
```

Write a program to sharpen the input image using:
1. Laplacian filter
2. Gradient operators of Sobel and Canny edge detector
3. Unsharp masking. To obtain blurred image, use the Gaussian filter of size 5x5
and sigma= 3

```python

#Laplacian Filter
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", 0)

kernel = np.array([[0,-1,0],
                   [-1,4,-1],
                   [0,-1,0]])

padded = np.pad(img.astype(float), 1, mode='constant')
result = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        total = 0

        for x in range(3):
            for y in range(3):
                total += padded[i+x,j+y] * kernel[x,y]

        value = img[i,j] + total
        result[i,j] = np.clip(value, 0, 255)

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")


plt.subplot(1,2,2)
plt.imshow(result, cmap='gray')
plt.title("Laplacian Sharpening")
plt.axis("off")


plt.show()
```

```python
#Sobel operator

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", 0)

gx = np.array([[-1,0,1],
               [-2,0,2],
               [-1,0,1]])

gy = np.array([[-1,-2,-1],
               [0,0,0],
               [1,2,1]])

padded = np.pad(img.astype(float), 1, mode='constant')
result = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        sx = 0
        sy = 0

        for x in range(3):
            for y in range(3):
                sx += padded[i+x,j+y] * gx[x,y]
                sy += padded[i+x,j+y] * gy[x,y]

        magnitude = np.sqrt(sx**2 + sy**2)
        value = img[i,j] + magnitude
        result[i,j] = np.clip(value, 0, 255)

plt.subplot(1,2,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(result, cmap='gray')
plt.title("Sobel Sharpening")
plt.axis("off")
plt.show()
```

```python
#Canny edge detector

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", 0)

edges = cv2.Canny(img, 100, 200)

result = np.zeros(img.shape, dtype=np.uint8)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        value = int(img[i,j]) + int(edges[i,j])
        result[i,j] = np.clip(value, 0, 255)

plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(edges, cmap='gray')
plt.title("Canny Edges")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(result, cmap='gray')
plt.title("Canny Sharpening")
plt.axis("off")

plt.show()
```

```python
#Unsharp masking

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("input.jpg", 0)

n = 5
sigma = 3
p = n // 2

padded = np.pad(img.astype(float), p, mode='constant')

kernel = np.zeros((n,n), dtype=float)

for x in range(n):
    for y in range(n):
        a = x-p
        b = y-p
        kernel[x,y] = np.exp(-(a*a+b*b)/(2*sigma*sigma))

kernel = kernel / np.sum(kernel)

blurred = np.zeros(img.shape, dtype=float)

for i in range(img.shape[0]):
    for j in range(img.shape[1]):
        total = 0

        for x in range(n):
            for y in range(n):
                total += padded[i+x,j+y] * kernel[x,y]

        blurred[i,j] = total

mask = img.astype(float) - blurred

sharpened = img.astype(float) + mask

sharpened = np.clip(sharpened, 0, 255).astype(np.uint8)
blurred = np.clip(blurred, 0, 255).astype(np.uint8)

plt.subplot(1,3,1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(blurred, cmap='gray')
plt.title("Gaussian Blurred")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(sharpened, cmap='gray')
plt.title("Unsharp Masking")
plt.axis("off")

plt.show()
```

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt


img = cv2.imread('input.jpg', 0)

if img is None:
    print("Image not found")
    exit()

img = img.astype(float)


gaussian_noise = np.random.normal(0, 20, img.shape)
gaussian_img = np.clip(img + gaussian_noise, 0, 255)

speckle_noise = np.random.normal(0, 0.2, img.shape)
speckle_img = np.clip(img + img * speckle_noise, 0, 255)



def wiener_filter(image, size=5):
    pad = size // 2
    padded = np.pad(image, pad, mode='reflect')
    result = np.zeros_like(image)

    noise_var = np.var(image - np.mean(image))

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            window = padded[i:i+size, j:j+size]

            mean = np.mean(window)
            var = np.var(window)

            result[i, j] = mean + max(0, var-noise_var) / (var+1e-8) * (image[i, j]-mean)

    return np.clip(result, 0, 255)



def median_filter(image, size=5):
    pad = size // 2
    padded = np.pad(image, pad, mode='reflect')
    result = np.zeros_like(image)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            window = padded[i:i+size, j:j+size]
            result[i, j] = np.median(window)

    return result



def gaussian_filter(image, size=5, sigma=1):
    pad = size // 2
    padded = np.pad(image, pad, mode='reflect')
    result = np.zeros_like(image)


    kernel = np.zeros((size, size))

    for i in range(size):
        for j in range(size):
            x = i - pad
            y = j - pad
            kernel[i, j] = np.exp(-(x*x+y*y)/(2*sigma*sigma))

    kernel = kernel / np.sum(kernel)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            window = padded[i:i+size, j:j+size]
            result[i, j] = np.sum(window * kernel)

    return np.clip(result, 0, 255)



def bilateral_filter(image, size=5, sigma_s=2, sigma_r=30):
    pad = size // 2
    padded = np.pad(image, pad, mode='reflect')
    result = np.zeros_like(image)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):

            window = padded[i:i+size, j:j+size]

            spatial = np.zeros((size, size))
            for x in range(size):
                for y in range(size):
                    dx = x - pad
                    dy = y - pad
                    spatial[x, y] = np.exp(
                        -(dx*dx + dy*dy)/(2*sigma_s*sigma_s)
                    )

            intensity = np.exp(
                -((window - image[i, j])**2)/(2*sigma_r*sigma_r)
            )

            weights = spatial * intensity

            result[i, j] = np.sum(weights * window) / np.sum(weights)

    return np.clip(result, 0, 255)


g1 = wiener_filter(gaussian_img)
g2 = median_filter(gaussian_img)
g3 = gaussian_filter(gaussian_img)
g4 = bilateral_filter(gaussian_img)

s1 = wiener_filter(speckle_img)
s2 = median_filter(speckle_img)
s3 = gaussian_filter(speckle_img)
s4 = bilateral_filter(speckle_img)

def display(images, titles):
    plt.figure(figsize=(12, 8))

    for i in range(len(images)):
        plt.subplot(2, 3, i+1)
        plt.imshow(images[i], cmap='gray')
        plt.title(titles[i])
        plt.axis('off')

    plt.tight_layout()
    plt.show()


titles = ['Original', 'Noisy', 'Wiener',
          'Median', 'Gaussian', 'Bilateral']

display([img, gaussian_img, g1, g2, g3, g4], titles)

display([img, speckle_img, s1, s2, s3, s4], titles)
```

 Write a program to denoise the input image using adaptive median filter.
Input the initial window size and maximum size of window (Smax) from
the user. Compare the performance of median filter and adaptive median
filter, using the PSNR values.

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('input.jpg', 0)

if img is None:
    print("Image not found")
    exit()

img = img.astype(float)

Smin = int(input("Enter initial window size: "))
Smax = int(input("Enter maximum window size: "))

noisy = img.copy()
prob = 0.1

random = np.random.random(img.shape)
noisy[random < prob/2] = 0
noisy[random > 1-prob/2] = 255


def median_filter(image, size):
    pad = size // 2
    padded = np.pad(image, pad, mode='reflect')
    result = np.zeros_like(image)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            window = padded[i:i+size, j:j+size]
            result[i, j] = np.median(window)

    return result


def adaptive_median(image, Smin, Smax):
    pad = Smax // 2
    padded = np.pad(image, pad, mode='reflect')
    result = np.zeros_like(image)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):

            size = Smin
            pixel = image[i, j]

            while size <= Smax:
                p = pad
                start = p - size//2

                window = padded[
                    i+start:i+start+size,
                    j+start:j+start+size
                ]

                zmin = np.min(window)
                zmax = np.max(window)
                zmed = np.median(window)

                A1 = zmed - zmin
                A2 = zmed - zmax

                if A1 > 0 and A2 < 0:
                    B1 = pixel - zmin
                    B2 = pixel - zmax

                    if B1 > 0 and B2 < 0:
                        result[i, j] = pixel
                    else:
                        result[i, j] = zmed

                    break

                else:
                    size += 2

            else:
                result[i, j] = zmed

    return result


def calculate_psnr(original, filtered):
    mse = np.mean((original - filtered) ** 2)

    if mse == 0:
        return float('inf')

    return 10 * np.log10((255 ** 2) / mse)


median_img = median_filter(noisy, Smin)
adaptive_img = adaptive_median(noisy, Smin, Smax)

psnr_median = calculate_psnr(img, median_img)
psnr_adaptive = calculate_psnr(img, adaptive_img)

print("\nPSNR Comparison")
print("Median Filter PSNR:", psnr_median, "dB")
print("Adaptive Median PSNR:", psnr_adaptive, "dB")

images = [img, noisy, median_img, adaptive_img]
titles = ['Original Image', 'Noisy Image',
          'Median Filter', 'Adaptive Median Filter']

plt.figure(figsize=(10, 8))

for i in range(4):
    plt.subplot(2, 2, i+1)
    plt.imshow(images[i], cmap='gray')
    plt.title(titles[i])
    plt.axis('off')

plt.tight_layout()
plt.show()
```

Write a program to threshold the image using Otsu's global threshold for
segmentation of input gray scale image

```python

import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread('input.jpg', 0)

if img is None:
    print("Image not found")
    exit()

hist, bins = np.histogram(img.flatten(), 256, [0, 256])

total = img.size
sum_total = np.sum(np.arange(256) * hist)

sum_bg = 0
weight_bg = 0
max_variance = 0
threshold = 0

for t in range(256):
    weight_bg += hist[t]

    if weight_bg == 0:
        continue

    weight_fg = total - weight_bg

    if weight_fg == 0:
        break

    sum_bg += t * hist[t]

    mean_bg = sum_bg / weight_bg
    mean_fg = (sum_total - sum_bg) / weight_fg

    variance = weight_bg * weight_fg * (mean_bg - mean_fg) ** 2

    if variance > max_variance:
        max_variance = variance
        threshold = t

print("Optimal Threshold:", threshold)

segmented = np.zeros_like(img)
segmented[img > threshold] = 255

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(segmented, cmap='gray')
plt.title("Otsu Segmentation")
plt.axis('off')

plt.tight_layout()
plt.show()
```
