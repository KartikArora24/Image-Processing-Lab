```python
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)
```

```python
_, binary = cv2.threshold(img, 127, 255, cv2.THRESH_BINARY)
plt.imshow(binary, cmap='gray')
plt.title("Binary Image")
plt.axis('off')
plt.show()
```

```python
complement = 255 - binary
plt.imshow(complement, cmap='gray')
plt.title("Binary Image Complement")
plt.axis('off')
plt.show()
```

```python
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)
img_float = img.astype(float)
c=256/np.log(257)
log_image = c*np.log1p(img_float)
log_image = cv2.normalize(log_image, None, 0, 255, cv2.NORM_MINMAX)
log_image = log_image.astype(np.uint8)
plt.imshow(log_image, cmap='gray')
plt.title("Logarithmic Image")
plt.axis('off')
plt.show()
```

```python
gamma = float(input("Enter the value of gamma: "))
c=1
normalized = img / 255.0
gamma_image = c * (normalized ** gamma)
gamma_image = np.uint8(gamma_image * 255)
plt.figure(figsize=(12,6))
plt.imshow(img, cmap='gray')
plt.title("Original Image")
plt.axis('off')
plt.show()
plt.imshow(gamma_image, cmap='gray')
plt.title("Power Law / Gamma Transformed Image")
plt.axis('off')
plt.show()
```

```python
r1 = int(input("Enter input range start: "))
r2 = int(input("Enter input range end: "))
s1 = int(input("Enter output range start: "))
s2 = int(input("Enter output range end: "))
```

```python
output = img.copy().astype(float)
mask = (img >= r1) & (img <= r2)
output[mask] = s1 + ((img[mask] - r1) * (s2 - s1) / (r2 - r1))
output = np.clip(output, 0, 255).astype(np.uint8)
```

```python
plt.imshow(output, cmap='gray')
plt.title("Contrast Stretched Image")
plt.axis('off')
plt.show()
```
